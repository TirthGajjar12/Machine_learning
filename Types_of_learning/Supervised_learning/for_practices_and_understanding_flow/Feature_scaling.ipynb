```python
#columns Transformer
import numpy as np 
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x1_train,x1_test,y1_test,y1_train = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2)
```

```python
print(x1_train)
```

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers=[('tnf1',SimpleImputer(),['fever']),('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),('tnf3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])],remainder='passthrough')
```

```python
transformer.fit_transform(x1_train)
```
