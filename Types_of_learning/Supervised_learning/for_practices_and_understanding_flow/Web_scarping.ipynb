```python
from bs4 import BeautifulSoup
import requests
import pandas as pd

url = "https://www.ambitionbox.com/list-of-companies"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, "html.parser")

companies = []

for card in soup.find_all("h2"):
    try:
        name = card.text.strip()

        parent = card.find_parent()

        companies.append({
            "Company": name
        })

    except Exception as e:
        print(e)

df = pd.DataFrame(companies)

print(df.head())
```

```python
import pandas as pd

data = []

cards = soup.find_all(
    "div",
    attrs={"itemprop":"itemListElement"}
)

for card in cards:

    company = card.find(
        "meta",
        {"itemprop":"name"}
    )["content"]

    url = card.find(
        "meta",
        {"itemprop":"url"}
    )["content"]

    data.append({
        "Company": company,
        "URL": url
    })

df = pd.DataFrame(data)

print(df.head())
```
